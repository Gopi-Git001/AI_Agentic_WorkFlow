```python
import os 

from dotenv import load_dotenv

from openai import OpenAI

load_dotenv(override=True)


openai = OpenAI(base_url=os.getenv("OLLAMA_OPENAI_BASE_URL"), api_key=os.getenv("MY_API_KEY"))


messages = [
    {"role": "system", "content": "You are a helpful assistant that can answer questions and help with tasks."},
    {"role": "user", "content": "Hi, My name is Gopi"}
]

model = os.getenv("DEFAULT_MODEL")

response = openai.chat.completions.create(
    model=model,
    messages=messages,
)


content = response.choices[0].message.content

print(content)
```
