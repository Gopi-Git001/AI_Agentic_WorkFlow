```python
import os 
from dotenv import load_dotenv

load_dotenv(override=True)
api_key = os.getenv("OPENAI_API_KEY")
if api_key is None:
    raise ValueError("OPENAI_API_KEY is not set in the environment variables")
else:
    print("OPENAI_API_KEY is set in the environment variables")
```

```python
import requests

headers = {"Authorization": f"Bearer {api_key}","content-type": "application/json"}

payload = {
    "model": "gpt-4o-mini",
    "messages": [{"role": "user", "content": "Hello, how are you?"}],
    "temperature": 0.7,
    "max_tokens": 100,
    "top_p": 1
}

payload
```

```python
from requests import post

response = requests.post("https://api.openai.com/v1/chat/completions", headers=headers, json=payload )
response.json()
```

```python
response.json()["choices"][0]["message"]["content"]
```

```python
from openai import OpenAI

load_dotenv(override=True)

GEMINI_BASE_URL = os.getenv("GEMINI_BASE_URL")

gemini_api_key = os.getenv("GEMINI_API_KEY")



gemini = OpenAI(base_url=GEMINI_BASE_URL, api_key=gemini_api_key)
client =gemini.chat.completions.create(
    model="models/gemini-2.5-flash",
    messages=[{"role": "user", "content": "Hello, how are you?"}],
)

client.choices[0].message.content

```

```python
# List available models to see the exact ID needed
models = gemini.models.list()
for model in models:
    print(model.id)
```

```python
load_dotenv(override=True)
ollama_api_key = os.getenv("ollama_api_key")

ollama_base_url = os.getenv("OLLAMA_OPENAI_BASE_URL")
ollama_client = OpenAI(base_url=ollama_base_url , api_key=ollama_api_key)
model = os.getenv("DEFAULT_MODEL")
response = ollama_client.chat.completions.create(
    model=model,
    messages=[{"role": "user", "content": "Hello, how are you?"}],
    temperature=0.7,
    max_tokens=100,
    top_p=1
)
response.choices[0].message.content

```
